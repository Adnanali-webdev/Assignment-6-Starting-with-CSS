Q1) Create a simple page with some div tags and show different ways to add CSS as well as what happens when you target the same elements with inline,internal, and external CSS. Also, utilize comments in the project where required

```python
from IPython.display import HTML

def css_styling():
    styles = open("style.css", "r").read()
    return HTML(f"<style>{styles}</style>")

css_styling()

html_code = """

<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Element target with Inline/Internal/External CSS</title>
    <!--    Internal CSS   -->
    <style>
        .box{
            background-color: darkkhaki;
            color:floralwhite;
            margin:20px;
            padding:10px;
        }
    </style>
    <!--External CSS -->
    <link rel="stylesheet" href="/content/style.css">
</head>
<body>

    <!--  inline CSS  style overrides Internal and External style  -->
    <div class="box" style="background-color:darkcyan; color:orangered">
        This div has Inline CSS
    </div>
     <!--Internal style applies unless overridden externally-->
    <div class="box">
        This div use Internal CSS and External may override it
    </div>
    <!--External style tries to override internal but has less priority-->
    <div class="box external">
        This div shows external CSS in action
    </div>


</body>
</html>
"""


display(HTML(html_code))

```

Q2) Build an HTML page with multiple paragraphs, each assigned a unique class name. Write CSS rules using class selectors to apply distinct styling to each paragraph. Follow the BEM naming convention and explain how you've named the classes

```python
from IPython.display import HTML

html_code = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>BEM Paragraph Styling</title>
    <link rel="stylesheet" href="styles.css">
    <style>
    /* Block base style */
.text-block {
    max-width: 600px;
    margin: auto;
    padding: 20px;
    font-family: Arial, sans-serif;
}

/* Element base style */
.text-block__paragraph {
    font-size: 16px;
    line-height: 1.6;
    margin-bottom: 15px;
}

/* Modifiers for different paragraph styles */
.text-block__paragraph--primary {
    color: white;
    background-color: #007BFF;
    padding: 10px;
    border-radius: 5px;
}

.text-block__paragraph--secondary {
    color: #333;
    background-color: #f2f2f2;
    padding: 10px;
    border-left: 5px solid #ccc;
}

.text-block__paragraph--warning {
    color: #856404;
    background-color: #fff3cd;
    padding: 10px;
    border-left: 5px solid #ffecb5;
}

.text-block__paragraph--success {
    color: #155724;
    background-color: #d4edda;
    padding: 10px;
    border-left: 5px solid #c3e6cb;
}
</style>
</head>
<body>

    <div class="text-block">
        <p class="text-block__paragraph text-block__paragraph--primary">
            This is the primary styled paragraph.
        </p>
        <p class="text-block__paragraph text-block__paragraph--secondary">
            This is the secondary styled paragraph.
        </p>
        <p class="text-block__paragraph text-block__paragraph--warning">
            This is the warning styled paragraph.
        </p>
        <p class="text-block__paragraph text-block__paragraph--success">
            This is the success styled paragraph.
        </p>
    </div>

</body>
</html>
"""
display(HTML(html_code))
```
